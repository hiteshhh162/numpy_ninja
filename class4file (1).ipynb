##Fancy Indexing
you can Pick elements from specific positions using a list/array of indices.

```python
import numpy as np
arr = np.array([10,20,30,40,50])
arr[[0,2,1]]
```

You can do this fancy indexing in 2D Matrix aswell

```python
matrix = np.arange(1,21).reshape(4,5)
print(matrix)
matrix[[3,2,0],[4,2,4]]
```

##Conditional Selection
Find values based on conditions or replace them.

```python
arr = np.arange(1,21)
np.where(arr %2==0,arr,0)
```

##Sorting & Ordering
np.sort() → returns sorted array

np.argsort() → returns indices of sorted order

```python
arr = np.array([1,2,1,4,2,4,7,5,4,7,5,8,58,58,5,854,85,8,658,32])
np.sort(arr)
```

```python
np.argsort(arr)
```

##Concatenation & Stacking

```python
a = np.array([1,2,3])
b=np.array([4,5,6])
np.concat([a,b])

```

Vertical & Horizontal Stacking

```python
a = np.array([[1,2,3],[4,5,6]])
b=np.array([[7,8,9],[10,11,12]])

np.vstack((a,b))
```

```python
np.hstack((a,b))
```
